## Validation Protocol
- Repeated Stratified 5-Fold Cross-Validation (10 repeats = 50 folds per model).
- Position-stratified splits on verified $N = 149$ eligible 2023 rookie cohort.
- In-fold median imputation and standard scaling fitted strictly on training sets.

```python
import sys
from pathlib import Path
import pandas as pd
import plotly.express as px

PROJECT_ROOT = Path("..").resolve()
metrics_df = pd.read_csv(PROJECT_ROOT / "models" / "saved" / "model_comparison_metrics.csv")
metrics_df
```

## 1. Out-of-Fold R² Comparison with 95% Confidence Intervals

```python
fig = px.bar(
    metrics_df,
    x="model_name",
    y="oof_r2",
    error_y=metrics_df["std_fold_r2"],
    color="model_name",
    title="Out-Of-Fold R² Comparison (50 Folds Repeated Stratified CV)",
    labels={"oof_r2": "Out-Of-Fold R²", "model_name": "Model Pipeline"},
    template="plotly_dark",
)
fig.show()
```
